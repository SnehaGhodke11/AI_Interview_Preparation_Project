```python
import pandas as pd

# Load dataset
df = pd.read_csv("AI_Career_Prediction_Dataset_320_Rows.csv")

# Show first 5 rows
df.head()
```

```python
# Dataset shape (rows, columns)
print("Shape:", df.shape)

# Column names
print("\nColumns:")
print(df.columns.tolist())

# Information about data types
print("\nDataset Info:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Features (Input)
X = df.drop("Career", axis=1)

# Target (Output)
y = df["Career"]

# Check
print("X Shape:", X.shape)
print("y Shape:", y.shape)

print("\nFirst 5 Careers:")
print(y.head())
```

```python
from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder
label_encoder = LabelEncoder()

# Convert career names into numbers
y_encoded = label_encoder.fit_transform(y)

# Check first 10 encoded values
print("First 10 Encoded Labels:")
print(y_encoded[:10])

# See career-to-number mapping
career_mapping = dict(zip(label_encoder.classes_, range(len(label_encoder.classes_))))
print("\nCareer Mapping:")
print(career_mapping)
```

```python
from sklearn.model_selection import train_test_split

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)

# Check the split
print("Training Features:", X_train.shape)
print("Testing Features :", X_test.shape)
print("Training Labels  :", y_train.shape)
print("Testing Labels   :", y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

# Create the model
model = DecisionTreeClassifier(random_state=42)

# Train the model
model.fit(X_train, y_train)

print("✅ Decision Tree Model Trained Successfully!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Predict on test data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.2%}")

# Detailed performance
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))
```

```python
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.model_selection import GridSearchCV

extra = ExtraTreesClassifier(
    random_state=42,
    class_weight="balanced"
)

param_grid_extra = {
    'n_estimators': [200, 300, 500],
    'max_depth': [10, 15, 20, None],
    'min_samples_split': [2, 3, 5],
    'min_samples_leaf': [1, 2],
    'max_features': ['sqrt', 'log2', None]
}

grid_extra = GridSearchCV(
    estimator=extra,
    param_grid=param_grid_extra,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1
)

grid_extra.fit(X_train, y_train)

best_extra = grid_extra.best_estimator_

print("\nBest Extra Trees Parameters:")
print(grid_extra.best_params_)

print("\nBest Extra Trees CV Accuracy:")
print(f"{grid_extra.best_score_:.2%}")
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred_extra = best_extra.predict(X_test)

accuracy_extra = accuracy_score(y_test, y_pred_extra)

print("\nExtra Trees Test Accuracy:", f"{accuracy_extra:.2%}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_extra,
    target_names=label_encoder.classes_
))
```

```python
# Keep the better model
best_model = best_extra

print("Final selected model: Extra Trees")
print(f"Test Accuracy: {accuracy_extra:.2%}")
```

```python
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import accuracy_score, classification_report
import joblib

# Final model
final_model = ExtraTreesClassifier(
    n_estimators=500,
    max_depth=15,
    max_features=None,
    min_samples_leaf=1,
    min_samples_split=5,
    random_state=42,
    n_jobs=-1
)

# Train final model
final_model.fit(X_train, y_train)

# Test
y_pred = final_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("====================================")
print("FINAL CAREER RECOMMENDATION MODEL")
print("====================================")
print(f"Model: Extra Trees")
print(f"Test Accuracy: {accuracy:.2%}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=label_encoder.classes_
))

# Save
joblib.dump(final_model, "career_recommendation_model.pkl")
joblib.dump(label_encoder, "label_encoder.pkl")

print("\nModel saved successfully!")
```

```python
import joblib
import pandas as pd

model = joblib.load("career_recommendation_model.pkl")
encoder = joblib.load("label_encoder.pkl")

features = [
    "Python",
    "Java",
    "SQL",
    "HTML",
    "C++",
    "Problem_Solving",
    "DBMS",
    "JavaScript",
    "Data_Structures",
    "Communication"
]

# Student A - Python focused
student_a = pd.DataFrame([[
    5, 1, 2, 3, 1, 5, 3, 2, 5, 3
]], columns=features)

# Student B - Web focused
student_b = pd.DataFrame([[
    1, 1, 2, 5, 1, 3, 2, 5, 2, 4
]], columns=features)

# Student C - Database focused
student_c = pd.DataFrame([[
    1, 2, 5, 2, 1, 3, 5, 1, 3, 4
]], columns=features)


for name, student in [
    ("Python Student", student_a),
    ("Web Student", student_b),
    ("Database Student", student_c)
]:

    prediction = model.predict(student)

    career = encoder.inverse_transform(prediction)[0]

    probabilities = model.predict_proba(student)[0]

    top_indices = probabilities.argsort()[-3:][::-1]

    print("\n==============================")
    print(name)
    print("==============================")
    print("Prediction:", career)

    for i in top_indices:
        career_name = encoder.inverse_transform([i])[0]
        print(
            f"{career_name}: "
            f"{probabilities[i]:.2%}"
        )
```
